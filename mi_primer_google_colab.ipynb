# HTML con Python
En el mundo real utilizaríamos librerías como Flask o Django para crear paginas web con Python. Pero desde colab vamos a hacerlo con la librería iPython que ya viene preinstalada en el propio Colab.

 IPython.display muestra contenido en formato HTML directamente en la salida de Colab



```python
# Utilizar recursos HTML de la librería IPhython
from IPython.display import display, HTML

#Contenido del HTML (lo almacenamos en una variable)
miContenido =f"""
<h1>Hola Mundo desde Python 🐍</h1>
<p>Soy un texto de un parrafo y <a href='https://google.com'>esto un enlace</a> en HTML.</p>
<img src='https://picsum.photos/id/33/800/200' alt='Texto alternativo'>
"""
display(HTML(miContenido))
```

# Ejemplo 03
Mezclamos Array alumnos con HTML y CSS

```python
# Utilizar recursos HTML de la librería IPhython
from IPython.display import display, HTML

#Array con listado de alumnos
datos =[
    {"nombre":"Miguel",    "apellido":"Lopez",      "annio":1983, "emoji":"🍕",  "id":22 },
    {"nombre":"Juan",      "apellido":"García",     "annio":1990, "emoji":"🍟",  "id":33 },
    {"nombre":"Pedro",     "apellido":"Antuña",     "annio":1995, "emoji":"🍔",  "id":44 },
    {"nombre":"Maria",     "apellido":"Carbajal",   "annio":2000, "emoji":"🥗",  "id":55 },
    {"nombre":"Ana",       "apellido":"Uría",       "annio":2005, "emoji":"🍖",  "id":66 },
]

# Almaceno variable con lo que voy a imprimir (miHTML)
fichas='';

# Bucle que recorre el array
for alumno in datos:
  fichas+=f"""
  <li>
  <img src='https://picsum.photos/id/{alumno['id']}/200/80' alt='Texto alternativo'>
  <h2>{alumno['nombre']}  {alumno['apellido']}</h2>
  <p>tengo {2026 - alumno['annio']} años. me gusta {alumno['emoji']}</p>
  """


#Contenido del HTML (lo almacenamos en una variable)
miContenido =f"""
<style>

h1, h2, h3, main{{
    color:black
}}

img{{
    max-width:100%;
}}

  main{{
    background:white;
    padding:6px;
    border-radius:20px;
    max-width:650px;
    margin:40px auto;

  }}

  .fichas{{
      list-style:none;
      padding-left:0;
      display:flex;
      gap:6px;
      flex-wrap:wrap;
  }}

  .fichas li{{
      flex:1 1 0;
      max-width:180px;
      border:solid 1px black;
      padding:4px;
      border-radius:8px;
  }}

</style>

<main>
<h1>Hola Mundo desde Python 🐍</h1>
<p>Soy un texto de un parrafo y <a href='https://google.com'>esto un enlace</a> en HTML.</p>
<img src='https://picsum.photos/id/33/800/200' alt='Texto alternativo'>
<ul>
  <li>Perro</li>
  <li>Gato</li>
  <li>Conejo</li>
</ul>

<ul class='fichas'>
{fichas}
</ul>

</main>
"""

#Mostrar el contenido del HTML
display(HTML(miContenido))
```
